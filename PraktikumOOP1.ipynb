```python
class Kucing:
 # Constructor
 def __init__(self, nama, warna):
  self.nama = nama # Attribute
  self.warna = warna # Attribute

 # Method
 def suara(self):
    return f"{self.nama} berkata: Meow!"

# Membuat Object (Instansiasi)
kucing1 = Kucing("Luna", "Putih")
kucing2 = Kucing("Garfield", "Oranye")
# Mengakses Attribute dan Method
print(kucing1.warna) # Output: Putih
print(kucing1.nama) # Output: Luna
print(kucing1.suara()) # Output: Luna berkata: Meow!
print(kucing2.warna) # Output: Oranye
print(kucing2.nama) # Output: Garfield
print(kucing2.suara()) # Output: Garfield berkata: Meow!
```

```python
# Parent Class
class Hewan:
    def makan(self):
        print("Hewan ini sedang makan.")


class Burung(Hewan):
    def terbang(self):
        print("Burung sedang terbang.")


# Tambahkan class Mamalia yang mewarisi dari Hewan
class Mamalia(Hewan):
    def menyusui(self):
        print("Mamalia sedang menyusui.")

    def berlari(self):
        print("Mamalia sedang berlari.")


class Kucing(Mamalia):
    def makan(self):
        print("Kucing sedang makan.")


# Instansiasi

# Object
beo = Burung()
beo.makan()  # Mewarisi dari Hewan (Output: Hewan ini sedang makan.)
beo.terbang()  # Method milik sendiri

kucing = Kucing()
kucing.makan()
kucing.menyusui() # Kucing adalah Mamalia, jadi bisa menyusui

unta = Mamalia() # Buat instance dari Mamalia
unta.berlari() # Sekarang unta bisa berlari
```

```python
class AkunBank:
    def __init__(self, saldo):
        self.__saldo = saldo  # Private attribute

    def cek_saldo(self):
        return self.__saldo

    def tarik(self, jumlah):
        if jumlah > 0:
            self.__saldo -= jumlah
        else:
            print("Jumlah tarik tunai harus lebih dari 0.")

    def setor(self, jumlah):
        if jumlah > 0:
            self.__saldo += jumlah
        else:
            print("Jumlah setor harus lebih dari 0.")


rekening = AkunBank(1000)
rekening.setor(500)
rekening.tarik(200)
print(rekening.cek_saldo())  # Output: 1300 (1000 + 500 - 200)
# print(rekening.__saldo)  # Ini akan Error! Tidak bisa akses langsung.
```

```python
class Hero:

    def __init__(self, nama, hp_awal, attack_power=10):
        self.nama = nama
        # Enkapsulasi: HP bersifat Private (hanya bisa diakses di dalam class ini)
        self.__hp = hp_awal
        self.attack_power = attack_power

    # GETTER: Cara resmi melihat HP
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0  # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        # Pakai setter/getter bahkan di dalam class sendiri agar aman
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

    def serang(self, lawan):
        print(f"{self.nama} menyerang {lawan.nama}!")
        lawan.diserang(self.attack_power)


# Class Mage adalah anak dari class Hero
class Mage(Hero):

    def __init__(self, nama, hp_awal, attack_power, mana):
        # Memanggil constructor milik Parent (Hero)
        super().__init__(nama, hp_awal, attack_power)
        self.mana = mana

    def info(self):
        print(
            f"{self.nama} [Mage] | HP: {self.get_hp()} | Mana: {self.mana}"
        )

    # Mage punya skill khusus
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.nama} menggunakan Fireball ke {lawan.nama}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)  # Damage 2x lipat
        else:
            print(f"{self.nama} gagal skill! Mana tidak cukup.")


# -- Main Program --
print("\n--- Uji Coba Enkapsulasi & Inheritance ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

# Uji Coba Info
eudora.info()

# Uji Coba Serangan & Skill
eudora.serang(balmond)  # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond)  # Skill khusus Mage

print("\n--- Uji Coba Setter & Cheat Protection ---")
hero1 = Hero("Layla", 100)
hero1.set_hp(-50)  # Coba set negatif -> HP diset ke 0
print(f"HP Layla setelah set -50 : {hero1.get_hp()}")

print(f"Mencoba akses paksa: {hero1._Hero__hp}")
hero1.set_hp(9999)
print(f"HP Layla setelah set 9999: {hero1.get_hp()}")
hero1.diserang(50) # Output: Layla terkena damage 50. Sisa HP: 50
hero1.diserang(500) # Output: Layla terkena damage 500. Sisa HP: 0
```

```python
from abc import ABC, abstractmethod


# 1. Interface / Abstract Class
# Ini adalah KONTRAK. Semua turunan wajib punya method di bawah ini.
class GameUnit(ABC):

    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass


# 2. Implementasi pada Class Konkret
class Hero(GameUnit):

    def __init__(self, nama):
        self.nama = nama

    # Kita WAJIB membuat method serang, kalau tidak akan Error
    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")


class Monster(GameUnit):

    def __init__(self, jenis):
        self.jenis = jenis

    # Implementasi serang versi Monster
    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")


# -- Uji Coba --
# unit = GameUnit() # ERROR! Abstract class tidak bisa jadi objek.

h = Hero("Alucard")
m = Monster("Serigala")

h.info()
m.info()
```

```python
# Parent Class
class Hero:

    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")


# Child Class 1
class Mage(Hero):

    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")


# Child Class 2
class Archer(Hero):

    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")


# Child Class 3
class Fighter(Hero):

    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")


# -- Penerapan Polymorphism --
# Daftar hero campuran
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord"),
]

print("--- PERANG DIMULAI ---")

# Satu perintah loop, tetapi respon tiap objek berbeda-beda (Polymorphism)
for pahlawan in pasukan:
    pahlawan.serang()
```
